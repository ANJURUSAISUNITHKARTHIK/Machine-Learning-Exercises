```
import pandas as pd
train_csv = pd.read_csv("train.csv")
train_csv.info()
```

```
train_csv['Cabin'] = train_csv['Cabin'].fillna('Unknown')
train_csv.isnull().sum()
```

```
train_csv['Age']= train_csv['Age'].fillna(train_csv['Age'].median())
train_csv["Embarked"] = train_csv["Embarked"].fillna(train_csv["Embarked"].mode()[0])
train_csv.isnull().sum()
```

```
X = train_csv.drop("Survived",axis=1)
X = X.drop(["PassengerId", "Name", "Ticket", "Cabin"], axis=1)
X = pd.get_dummies(X, drop_first=True)
X.head()
```

```
Y = train_csv["Survived"]
```

```
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
#X_train.shape
#X_test.shape
```
